## KNN Algorithm in Supervised ML
## KNeighborsClassifier of sklearn library for this one

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

```python
X = np.array([
    [1,3500],
    [2,3200],
    [3,3000],
    [4,2500],
    [5,2300],
    [6,2000],
    [7,2200],
    [2,3700]])
y = ['Unfit','Unfit','Unfit','Fit','Fit','Fit','Fit','Unfit']
```

```python
#Create Model
model=KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(X,y)
```

```python
#predict
new_person = [[4.5,2600]] #4.5hrs exercise,2600 cal
pred = model.predict(new_person)
print('Predicted Result : ',pred[0])
```

```python
#prediction from taking the valueas from user
exercise_hours=float(input('Enter Hours of Exercise : '))
cal_intake=float(input('Enter Amount of Calories Consumed'))
pred_fut=model.predict([[exercise_hours,cal_intake]])
print(f'Exercise Hours {exercise_hours}, Calories Intake {cal_intake}, You are likely {pred_fut[0]}')
```

## Example 2

```python
import pandas as pd
```

```python
df = pd.read_csv('fruit_data.csv')
df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df
```

```python
#split the data
X = df[['Weight','Size']]
y = df['Fruit']
```

```python
model=KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(X,y)
```

```python
#base on data provided
new_fruit=pd.DataFrame([[180,7.5]],columns=['Weight','Size'])
pred = model.predict(new_fruit)
print('Perdicted Result : ',pred[0])
```

```python
#User input Values
fruit_weight=float(input('Enter Fruit Weight : '))
fruit_size=float(input('Enter Fruit Size : '))
our_fruit=pd.DataFrame([[fruit_weight,fruit_size]],columns=['Weight','Size'])
pred_fruit=model.predict(our_fruit)
print(f' Fruit Weight {fruit_weight}gm, Fruit Size {fruit_size}cm,The fruit is most likely {pred_fruit[0]}')
```

## Assignment 1

```python
dv=pd.read_csv('student_grade.csv')
```

```python
dv
```

```python
X = dv[['Hours_Studied','Assignments_Completed']]
y = dv['Grade']
```

```python
model=KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(X,y)
```

```python
#User input Values
hours_study=float(input('Enter Hours of study : '))
assign_comp=float(input('Enter Assignmment done : '))
result_expect=pd.DataFrame([[hours_study,assign_comp]],columns=['Hours_Studied','Assignments_Completed'])
pred_result=model.predict(result_expect)
print(f' Hours Study {hours_study}, with Assignment {assign_comp} completed, most likely {pred_result[0]}')
```
