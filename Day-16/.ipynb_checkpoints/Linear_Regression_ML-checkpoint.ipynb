# Supervised Machine Learning-Linear Regression

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('sales_data.csv')
data.head()
```

```python
x = data[['Advertising_Spend']]
y = data['Sales_Revenue']
```

```python
model = LinearRegression()
model.fit(x,y)
print('Training Done')
```

```python
new_spend = int(input('Enter New Spend to Predict Sales Revenue'))
predicted_revenue = model.predict([[new_spend]])
print(f'New Spend {new_spend} Predicted Revenue {predicted_revenue.round(2)}') 
```
