# 2. Logistic Regression Pass/Fail

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv('pass_fail.csv')
display(df)
```

```python
X =df[['Hours']]
y =df['Result']
```

```python
model=LogisticRegression()
model.fit(X,y)
```

```python
pred_35 = model.predict([[3.5]])
print('Predicted result for 3.5 : ',pred_35[0])
```

```python
hours_study=float(input('Numbers of Hours studied'))
pred_value=model.predict([[hours_study]])
print(f'After {hours_study} hours study you are likely to {pred_value[0]}')
```

## example 2

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv('customer_purchase.csv')
data
```

```python
le=LabelEncoder()
data['Purchased'] = le.fit_transform(data['Purchased'])
X = data[['Age','Salary']]
y = data['Purchased']
data
```

```python
model=LogisticRegression()
model.fit(X,y)
```

```python
#predict the result
new_customer = [[28, 35000]]
prediction = model.predict(new_customer)
print(f"Predicted Result (1=Yes, 0=No): {prediction[0]}")
```

```python
new_cust_age=int(input('Enter Age : '))
new_cust_sal=int(input('Enter Salary : '))
new_customer=[[new_cust_age,new_cust_sal]]
prediction=model.predict(new_customer)
if (prediction):
    print('Purchased')
else :
    print('Not Purchased')
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import CountVectorizer
```

```python
dls = pd.read_csv('spam_dataset.csv')
dls.head()
```

```python
dls.isnull().sum() 
```

```python
cv=CountVectorizer()
```

```python
X = cv.fit_transform(dls[['message_content']])
y = dls['is_spam']
```

```python
model = LogisticRegression()
```

```python
model.fit(X,y)
```

```python
new_msg=input('Enter your email here : \n')
new_msg_CV=cv.transform([new_msg])
predict_spam=model.predict(new_sg_CV)
if (predict_spam):
    print('This is spam message')
else :
    print('This is not a spam message')
```
