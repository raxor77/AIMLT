**We are going to write Multi Label Classification Example using TensorFlow**

```python
import tensorflow as tf
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
data={
    "overview":[
        "Friday the 13th",
        "American Horror Story",
        "The Dark Knight Rises",
        "The Emperor's New Groove",
        "Jurassic Park",
        "A Hero Saves the City from destruction",
        "A girl and boy become friends during war",
        "A funny sorry of collegues during office meet",
        "One Scary nights in a deserted place"

    ],
    "genres" :[["Horror"],["Drama","Horror"],["Action","Advanture"],["Comedy","Animation"],["Action","Advanture","Thriller"],["Action","Advanture"],["Romance","Drama"],["Comedy"],["Horror"]]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
tokenizer=Tokenizer(num_words=5000,oov_token="<OOV>")
tokenizer.fit_on_texts(df["overview"])
X=tokenizer.texts_to_sequences(df["overview"])
X=pad_sequences(X,maxlen=20,padding="post")
```

```python
mlb=MultiLabelBinarizer()
y=mlb.fit_transform(df["genres"])
```

```python
y
```

```python
#Create Model
model=tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=5000,output_dim=16,input_length=20),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(16,activation="relu"),
    tf.keras.layers.Dense(len(mlb.classes_),activation="sigmoid"),
])
```

```python
#Compile model
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
history = model.fit (X,y,epochs=20)
```

```python
# Evaluate Model
loss,accuracy = model.evaluate(X,y)
print("Test Loss :",loss)
print("Test Accuracy :",accuracy)
```

```python
#Prediction on a New Movie Overview
new_movie_overview = ['Adventure of Tom the cat']
seq=tokenizer.texts_to_sequences(new_movie_overview)
seq=pad_sequences(seq,maxlen=20,padding="post")
prediction=model.predict(seq)
prediction_labels=[mlb.classes_[i] for i in range(len(prediction[0])) if prediction[0][i] > 0.5]
print("Predicted Genres :",prediction_labels)
```

Use this data set to write one more MultiLabel Classification prediction using Nueral Network

```python
news_data = {
    "headline": [
        "Government launches new digital economy policy",
        "Star player wins championship after dramatic final",
        "Tech company releases AI-powered smartphone",
        "Stock market drops amid political tensions",
        "New environmental policy gains global support",
        "Space agency announces manned Mars mission"
    ],
    "topics": [
        ["Politics", "Economy", "Technology"],
        ["Sports"],
        ["Technology", "Business"],
        ["Economy", "Politics"],
        ["Environment", "Politics"],
        ["Science", "Technology"]
    ]
}
```

```python
ndf=pd.DataFrame(news_data)
ndf.head()
```

```python
tokenizer=Tokenizer(num_words=5000,oov_token="<OOV>")
tokenizer.fit_on_texts(ndf["headline"])
X=tokenizer.texts_to_sequences(ndf["headline"])
X=pad_sequences(X,maxlen=20,padding="post")
```

```python
mlbhl=MultiLabelBinarizer()
y=mlbhl.fit_transform(ndf["topics"])
```

```python
#Create Model
model_hl=tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=5000,output_dim=16,input_length=20),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(16,activation="relu"),
    tf.keras.layers.Dense(len(mlbhl.classes_),activation="sigmoid"),
])
```

```python
#Compile model
model_hl.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
history_hl = model_hl.fit (X,y,epochs=20)
```

```python
# Evaluate Model
loss,accuracy = model_hl.evaluate(X,y)
print("Test Loss :",loss)
print("Test Accuracy :",accuracy)
```

```python
#Prediction on a News topics
news_topics = ['Fire broke at Warehouse near Port in Lubenon']
seq_news=tokenizer.texts_to_sequences(news_topics)
seq_news=pad_sequences(seq_news,maxlen=20,padding="post")
prediction=model_hl.predict(seq_news)
prediction_topics=[mlbhl.classes_[i] for i in range(len(prediction[0])) if prediction[0][i] > 0.5]
print("Predicted News Topics :",prediction_topics)
```
