```python
import tensorflow as tf
from tensorflow.keras import layers
import pandas as pd,numpy as np,matplotlib.pyplot as plt
```

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()
```

```python
X_train,X_test=X_train/255.0,X_test/255.0
```

```python
model=tf.keras.Sequential([
    layers.Conv2D(32,(3,3),activation='relu',input_shape=(32,32,3)),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64,(3,3),activation='relu'),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(64,activation='relu'),
    layers.Dense(10,activation='softmax')
])
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
history = model.fit(X_train, y_train, epochs=2, validation_data=(X_test, y_test))
```

```python
loss,accuracy = model.evaluate(X_test,y_test)
print(f"Loss: {loss}")
print(f"Accuracy: {accuracy}")
```

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()
```

```python
X_train_small = X_train[:10000]
X_test_small = X_test[:10000]
y_train_small = y_train[:10000]
y_test_small = y_test[:10000]
```

```python
X_train_small,X_test_small=X_train_small/255.0,X_test_small/255.0
```

```python
model_small=tf.keras.Sequential([
    layers.Conv2D(32,(3,3),activation='relu',input_shape=(32,32,3)),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64,(3,3),activation='relu'),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(64,activation='relu'),
    layers.Dense(10,activation='softmax')
])
```

```python
model_small.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
history = model_small.fit(X_train_small, y_train_small, epochs=10, validation_data=(X_test_small, y_test_small))
```

```python
loss,accuracy = model_small.evaluate(X_test_small,y_test_small)
print("Test Accuracy",accuracy)
```
