# K-Means Clustring Example
# K-means comes under unsupervised Machine Language Technices
## Terms we will use here like cluster,centroid
### cluster : We will partition our datasets into K distinct cluster
### centroid : nearest mean cluster

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
X=np.array([
    [7.5,8.0],
    [9.0, 7.0],
    [10.5, 6.5],
    [6.0, 9.0],
    [13.0, 5.5],
    [8.0, 8.5],
    [14.0, 4.5],
    [9.5, 7.0],
    [7.0, 9.0],
    [12.0, 6.0]
])
df = pd.DataFrame(X, columns=['EntryTime','Duration'])
df
```

```python
#create model
model = KMeans(n_clusters=3,random_state=0)
```

```python
df['Cluster'] = model.fit_predict(df[['EntryTime','Duration']])
centers = model.cluster_centers_
df
```

```python
plt.figure(figsize=(6,5))
plt.scatter(df['EntryTime'], df['Duration'], c=df['Cluster'],s=50, marker='p')
plt.scatter(centers[:,0], centers[:,1], marker='x', s=100, c='red', label='Centroid')
plt.xlabel('Entry Time (Hour of Day)')
plt.ylabel('Duration in Office (Hours)')
plt.title('Working Hour Patterns using K-Means')
plt.legend()
plt.grid(True)
plt.show()
```

## Example 2

```python
#Load Data
data=pd.read_csv('customers.csv')
data.head()
```

```python
data.isnull().sum()
```

```python
#Create Cluster
model = KMeans(n_clusters=5,random_state=0)
```

```python
data['Cluster'] = model.fit_predict(data[['AnnualIncome','SpendingScore']])
centers = model.cluster_centers_
data.head()
```

```python
plt.figure(figsize=(6,5))
plt.scatter(data['AnnualIncome'], data['SpendingScore'], c=data['Cluster'],s=50, marker='p')
plt.scatter(centers[:,0], centers[:,1], marker='x', s=100, c='red', label='Centroid')
plt.xlabel('Anual Income')
plt.ylabel('Spending Score')
plt.title('High Income/High Spender')
plt.legend()
plt.grid(True)
plt.show()
```
