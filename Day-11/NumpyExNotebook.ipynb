### One Dimentional Array (1-D)

```python
import numpy as np
```

```python
array1=np.array([10,20,30,40,50])
```

```python
array1  #or can use print(array1)
```

### Two Dimentional Array (2-D)

```python
array_d2=np.array([[10,20,30],[40,50,60]])
```

```python
print(array_d2)
```

### Three Dimentional Array (3-D)

```python
array_d3=np.array([[[10,20,30],[40,50,60]],[[10,20,30],[40,50,60]]])
```

```python
print(array_d3)
```

### Find out Dimension of Numpy Array

```python
print('First Array Dimension :\t',array1.ndim)
```

```python
print('Second Array Dimension :\t',array_d2.ndim)
```

```python
print('Third Array Dimension :\t',array_d3.ndim)
```
