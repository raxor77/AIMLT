#### Searching and Sorting Example in 1-d Array

```python
our_arr=np.array([100,200,300,7000,400,500,300,1900,600])
```

```python
#print(np.argmax(our_arr))
```

```python
print(our_arr)
```

```python
sorted_arr=np.sort(our_arr)
print("sorted Arry: ")
print(sorted_arr)
```

```python
print('Is there any item greater than 500? :',np.any(sorted_arr>500))
```

```python
print('All greater than 500? :',np.where(sorted_arr>500))
```

```python
print('All Numbers greater than 500')
for num in np.where(sorted_arr>500):
    print(num)
```

```python
print(np.argmax(sorted_arr))
```

```python
selected_values=sorted_arr[sorted_arr>500]
```

```python
print(selected_values)
```

```python
for num in selected_values:
    print(num)
```

```python
for num in sorted_arr[sorted_arr>500]:
    print(num)
```

```python
print(' ',sorted_arr[sorted_arr>500])
```

## Slicing 1-D array
    1. to take elements from starting index to end index
    2. e.g.[start:end]

```python
print(our_arr)
```

```python
print(sorted_arr)
```

```python
slice_our_arr=our_arr[1:4]
print(slice_our_arr)
```

## Starting from an index and going till last
    ## [start:end]


```python
print(our_arr[3:])
```

```python
print(our_arr[:3])
```

```python
print(sorted_arr[:7])
```

```python
print(sorted_arr[5:])
```

```python
print(our_arr[:7])
```

## Step
    ## [start:end:step]
    ## step:every step element


```python
our_slice=sorted_arr[1:6:2]
print(our_slice)
```

```python
print('Original Arry')
print(sorted_arr)
print('Every Third Element')
every_third_element=sorted_arr[::3]
print(every_third_element)
```

### Slice Reverse

```python
print('Original Arry')
print(sorted_arr)
print('Reversel Arry : ')
print(sorted_arr[::-1])
```

```python
print('Last Five Elements : ')
print(sorted_arr[-5:])
```
