```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_excel('students.xlsx')
print(df)
```

```python
marks=df['Marks'].to_numpy()
print('Numpy Array of Marks')
print(marks)
```

```python
print('Highest Marks', np.max(marks))
```

```python
print('Lowest Marks', np.min(marks))
```

```python
print('Average Marks', np.mean(marks))
```

## .loc :label Based
## .iloc : index Based

```python
print('Original Dataframe')
print(df)
print('Location : Label Based')
print(df.loc[4])
print(df.loc[4,['Name','Age']])
```

```python
print('Original Dataframe')
print(df)
print('Location : Index Based')
print(df.iloc[4])
print(df.iloc[4,[0,2]])
```

## Axis in Numpy : 
#### 1.Axis 0 : Vertical direction- work column-wise
#### 2.Axis 1 : Horizontal direction- work row-wise

We have 2 students marks in 3 subjects
[[45,50,49],[30,45,49]]

```python
marks_arr=np.array([[45,50,49],[30,45,49]])
print(marks_arr)
```

```python
print('sum down each column',np.sum(marks_arr, axis=0))
```

```python
print('sum down each column',np.sum(marks_arr, axis=1))
```

#### find the student Name who got highest marks

```python
topper=df.loc[df['Marks'].idxmax(),'Name']
print('Top Student Name :',topper)
```

```python
lowest=df.loc[df['Marks'].idxmin(),'Name']
print('Top Student Name :',lowest)
```

```python
highest=df.iloc[df['Marks'].idxmax(),0]
print('Top Student Name :',highest)
```

```python
bottom=df.iloc[df['Marks'].idxmin(),0]
print('Top Student Name :',bottom)
```
