```python
# import required libraries
# NLTK is Naturel Language Tool Kit
import nltk
import spacy
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
import re
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
# import string
# import warnings
# warnings.filterwarnings('ignore')
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
our_text="Artifical Intelligence and Machine Learning are going to change the future of technology!"
```

```python
#split text into individual words/token
tokens=nltk.word_tokenize(our_text)
tokens
```

```python
# cleaning tokens
clean_tokens = [re.sub(r'[^a-zA-Z]', '', token.lower()) for token in tokens if token.isalpha()]
# clean_tokens = [token.lower() for token in tokens if token.isalpha()]  # <- same result
clean_tokens
```

```python
# Stop words Removal
stop_words = set(stopwords.words('english'))
filtered_tokens = [token for token in clean_tokens if token not in stop_words]
# Removes common less meaningful words
filtered_tokens
```

```python
#Steaming
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(token) for token in filtered_tokens]
#Reduce the words to their root form (not always a word from dictionary)
stemmed_tokens
```

```python
# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]
# convert words to their dictonary form (lemma) using POS : part of speech
lemmatized_tokens
```

Bag of words(BOW)

```python
corpus = ["Artificial Intelligence is transforming inductries",
          "Machine Learning is a branch of Artificial Intelligence",
          "Deep Larning is a subfield of Machine Learning",
          "Natural Language Processing is a subfield of Artificial Intelligence"]
```

```python
CV = CountVectorizer()
bow_matrix = CV.fit_transform(corpus)
print("BOW Matrix Array")
print(bow_matrix.toarray())
print("Feature Name BOW")
print(CV.get_feature_names_out())
```
