```python
import spacy
nlp=spacy.load("en_core_web_sm")
print('SpaCy Loadded Succcessfully')
```

```python
text =  "Bill Gates founded Microsoft in Year 1994 in California - USA"
```

```python
doc = nlp(text)
for ent in doc.ents:
  print(f'Text: {ent.text}, \nLabel: {ent.label_}, \nExplanation: {ent.text,ent.label_, spacy.explain(ent.label_)}\n')
```

```python
from spacy import displacy
displacy.render(doc, style='ent', jupyter=True)
```

```python
corpus = [
    "Tajmahal is located in Agra India ",
    "Google was founded by Larry Page in USA",
    "Apple was founded by Steve Jobs in USA",
    "Twin Towers is located in Kuala Lumpur",
]
```

```python
for text in corpus:
  doc = nlp(text)
  print(f"Text: {text}\n")
  for ent in doc.ents:
    print(f"{ent.text} --> {ent.label_}\n")
```
