# Label Encoding Basic

```python
from sklearn.preprocessing import LabelEncoder

colors=['Red','Blue','Green','Blue','Red','Green','Black','White']

le=LabelEncoder()
#fit and transform, auto assign numbers to label
encoded=le.fit_transform(colors)
print(encoded)
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

data=pd.DataFrame({'Color':['Red','Blue','Green','Blue','Red','Green','Black','White']})

#Create Encoder
ohe = OneHotEncoder(sparse_output=False)
#fit and transform, auto assign numbers to label
encoded = ohe.fit_transform(data[['Color']])
encoded_df = pd.DataFrame(encoded,columns=ohe.get_feature_names_out(['Color']))
print(encoded_df)
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

data=pd.DataFrame({'Region':['North','South','East','West','South']})

#Create Encoder
ohe = OneHotEncoder(sparse_output=False)
#fit and transform, auto assign numbers to label
encoded = ohe.fit_transform(data[['Region']])
encoded_df = pd.DataFrame(encoded,columns=ohe.get_feature_names_out(['Region']))
print(encoded_df)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

#step1 :collect and prepare data
hours_data=np.array([[1],[2],[3],[4],[5],[6],[7],[8]])
marks_data=np.array([25,35,45,50,55,65,70,80])
#step2 :create the model for Linear Regression
model=LinearRegression()
#step3 Train the model fit(X,y)
model.fit(X=hours_data,y=marks_data)
print('Model Train successful')
#step4 :predict marks
marks_pred=model.predict([[9]])
print(marks_pred)
```

```python
hours_data=float(input('Numbers of hours study: '))
pre_marks=model.predict([[hours_data]])
print(f'Predict Marks for {hours_data} hours is likely{pre_marks}')
```
