```python
import pandas as pd
```

```python
data={'students':['rahman','oon','razak','mahadhir','abdullah','najib'],'marks':[85,None,65,75,None,55]}
```

```python
df=pd.DataFrame(data)
df
```

```python
print('Print Missing data')
print('Numbers of missing values :',df.isnull().sum())
```

```python
avg_marks=df['marks'].mean()
print(avg_marks)
```

```python
df.fillna(avg_marks,inplace=True)
```

```python
print('after fillna() call')
df
```

## Lets see how we will use in our project

```python
students_data={'Names':['rahman','oon','razak','mahadhir','abdullah','najib'],'Marks':[85,None,65,75,None,55]}
df=pd.DataFrame(students_data)
print('Original Data')
df
```

```python
df.fillna(df['Marks'].mean(),inplace=True)
df
```

## dropna() example

```python
students_data={'Names':['rahman','oon','razak','mahadhir','abdullah','najib'],'Marks':[85,None,65,75,None,55]}
df=pd.DataFrame(students_data)
df
```

```python
df=df.dropna(subset=['Marks'])
```

```python
print('Print Dataset after dropna()')
df
```

## Another Exmaple to use

```python
emp_data={'Names':['rahman','oon','razak','mahadhir','abdullah','najib'],
          'Age':[35,None,25,20,24,None],
          'Salary':[50000,30000,20000,40000,None,None]}
```

```python
df=pd.DataFrame(emp_data)
print('Original Data')
df
```

```python
print('Numbers of missing values :',df.isnull().sum())
```

```python
df_drop=df.dropna()
df_drop
```

```python
df_drop_salary=df.dropna(subset=['Salary'])
df_drop_salary
```

```python
df.fillna({'Age':df['Age'].mean(),'Salary':df['Salary'].mean()}, inplace=True)
df
```
