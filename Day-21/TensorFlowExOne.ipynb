Random Tensor

```python
# Random Tensor
random_tensor=tf.random.Generator.from_seed(42)
random_tensor=random_tensor.normal(shape=(3,2))
random_tensor
```

```python
random_tensor_2=tf.random.uniform(shape=(300,150),minval=0,maxval=1)
random_tensor_2
```

```python
#same seed, same tensor
random_tensor_3=tf.random.Generator.from_seed(42)
random_tensor_3=random_tensor_3.normal(shape=(3,2))
random_tensor_4=tf.random.Generator.from_seed(42)
random_tensor_4=random_tensor_4.normal(shape=(3,2))
random_tensor_3,random_tensor_4
```

```python
#diff seed, diff tensor
random_tensor_3=tf.random.Generator.from_seed(42)
random_tensor_3=random_tensor_3.normal(shape=(3,2))
random_tensor_4=tf.random.Generator.from_seed(15)
random_tensor_4=random_tensor_4.normal(shape=(3,2))
random_tensor_3,random_tensor_4
```

```python
import numpy as np
numpy_Array=np.arange(1,25,dtype=np.int16)
numpy_Array
```

```python
tensor_numpy=tf.constant(numpy_Array,shape=(2,3,4))
tensor_numpy
```

```python
print("Tensor Information")
print("Shape:",tensor_numpy.shape)
print("Dimension:",tensor_numpy.ndim)
print("Rank :", tf.rank(tensor_numpy).numpy())
print("Size :", tf.size(tensor_numpy))
print("DataType of elements : ",tensor_numpy.dtype)
print("Elements along axis 0 : ",tensor_numpy.shape[0])
print("Elements along axis 1 : ",tensor_numpy.shape[1])
print("Elements along axis 2 : ",tensor_numpy.shape[2])
print("Total numbers of element : ",tf.size(tensor_numpy).numpy())
```

```python
# Working with item of each dimension
# First threeelements
tensor_numpy[:3,:3,:3]
```

```python
# First two elements of each dimension
tensor_numpy[:2,:2,:2]
```

** Arthmetic Operations with Tensors **
+Add
+Sub
+Multiply
+Div

```python
tensor_one=tf.constant([[11,9],[2,5]])
tensor_one
```

```python
#tensor_one+15
tf.add(tensor_one,15)    #also can use this
```

```python
# tensor_one-5
tf.subtract(tensor_one,5)   #also can use this
```

```python
# tensor_one*2
tf.multiply(tensor_one,2)   #also can use this
```

```python
# tensor_one/2
tf.divide(tensor_one,2)   #also can use this
```

```python
second_matrix=tf.constant([[8,16],[64,8.25],[625,9.23],[125,2.34],[169,144]],dtype=tf.float32)
print(second_matrix)
```

```python
#Squareroot only work on float dtype
tf.sqrt(second_matrix)
```
