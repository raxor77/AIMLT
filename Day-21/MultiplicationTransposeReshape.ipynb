**Matrix Multiplication**
# we will use tf.matmul() method
**Rule**
1. Inner dimension must match
  (m,n)@(n,p)
  eg.
  (5,7)@(8,9) It will not work bc n are diff
  (5,8)@(8,9) will work bc same inner demension.
2. The result matrix has the shape of outer demension.
meaning
A(m,n)
B(n,p)
so result has (m,p)
eg.
(4,7)@(7,5) ->(4,5)

```python
import tensorflow as tf
import numpy as np
```

```python
tensor_A=tf.constant([[1,2],[3,4],[4,5]],dtype=tf.int32)
tensor_B=tf.constant([[5,6],[7,8]],dtype=tf.int32)
print("Shape of A",tensor_A.shape)
print("Shape of B",tensor_B.shape)
```

```python
tensor_C=tf.matmul(tensor_A,tensor_B)
print(tensor_C)
print("Shape of C",tensor_C.shape)
```

```python
X=tf.constant([
               [1.,0.,-1.],
               [2.,3.,1.]
               ],
              dtype=tf.float32)
X
```

```python
Y=tf.constant([
               [4.],
               [5.],
               [6.]
               ],
              dtype=tf.float32)
Y
```

```python
print("X Shape",X.shape)
print("Y Shape",Y.shape)
```

```python
Z=tf.matmul(X,Y)
print(Z)
print("Z Shape",Z.shape)
```

```python
print(f'X-Shape : {X.shape},Y-Shape : {Y.shape}, Z-Shape : {Z.shape}')
print(X.numpy)
print(Y.numpy)
print(Z.numpy)
```

Transpose : Transpose swaps axes. For 2D matrix shape (m,n) become (n,m) after transpose

```python
tensor_one = tf.constant([[[1,2],[2,3]],[[4,5],[6,7]]],
                         dtype=tf.int32)
print(tensor_one)
print("Shape of tensor_one",tensor_one.shape)
```

```python
tensor_one_transpose=tf.transpose(tensor_one)
print(tensor_one_transpose)
print("Shape Tensor One Transpose",tensor_one_transpose.shape)
```

```python
# 3D Transpose with explicit axis permutation
tensor_two=tf.constant([
    [[1,2],[3,4]],
    [[5,6],[7,8]],
    ],dtype=tf.int32)
print(tensor_two)

```

```python
#swap first and last axes (0,1,2)->(0,2,1)
t2_transpose=tf.transpose(tensor_two,perm=[0,2,1])
print(t2_transpose)
print("Shape of t2_transpose",t2_transpose.shape)
print("Shape of tensor_two",tensor_two.shape)
```

**Reshape**
# tf.reshape(tensor,new_shape)
# change the shape without changing the data.
**Rule** The total number of elements must stay the same

```python
tensor=tf.constant([
                    [1,2,3],
                    [4,5,6]
                    ],dtype=tf.int32)
print(tensor)
print("Shape of tensor",tensor.shape)
print("Total number of elements in tensor",
      tf.size(tensor).numpy())
```

```python
tensor_r2=tf.reshape(tensor,[1,6])
print(tensor_r2)
print("Shape of tensor_r2",tensor_r2.shape)
print("Total number of elements in tensor_r2 are :",
      tf.size(tensor_r2).numpy())
```

```python
tensor_r3=tf.reshape(tensor,[6,1])
print(tensor_r3)
print("Shape of tensor_r3",tensor_r3.shape)
print("Total number of elements in tensor_r3 are :",
      tf.size(tensor_r3).numpy())
```

```python
#infer dimension
tensor_r4=tf.reshape(tensor,shape=[-1,2])
print(tensor_r4)
print("Shape of tensor_r4",tensor_r4.shape)
print("Total number of elements in tensor_r4 are :",
      tf.size(tensor_r4).numpy())
```
