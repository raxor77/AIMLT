```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('netflix_data.csv')
df.head(3)
```

```python
sns.set_theme(style='darkgrid',palette='Reds')
sns.countplot(data=df,x='type',hue='type')
plt.title('Number of Movies vs Number of TV Show')
plt.xlabel('Contents Type')
plt.ylabel('Count')
plt.show()
```

```python
df=df.dropna(subset=['genre'])
```

```python
plt.figure(figsize=(20,10))
sns.countplot(data=df,x='genre',hue='type',palette='bright')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.grid(axis='y',linestyle='--',color='lightblue',alpha=0.5)
plt.xticks(fontsize=15,color='darkblue',rotation=85)
#plt.tight_layout()
plt.show()
```

```python
df['genre']
```

```python
df=df.dropna(subset=['type','country'])
sns.countplot(data=df.head(10),x='type',hue='country',palette='deep')
plt.title('Netflix Contents Type by Country only top 10')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.grid(axis='y',linestyle='--',color='lightblue',alpha=0.5)
plt.xticks(fontsize=10,color='darkblue',rotation=30)
plt.tight_layout()
plt.show()
```

```python
df=df.dropna(subset=['type','country'])
plt.figure(figsize=(15,20))
sns.countplot(data=df,x='type',hue='country',palette='deep')
plt.title('Netflix Contents Type by Country only top 10')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.grid(axis='y',linestyle='--',color='lightblue',alpha=0.5)
plt.legend(title='Country',bbox_to_anchor=(1.02,1),loc='upper left')
#plt.tight_layout()
plt.show()
```

## Use Netflix dataset tovisualize the distribution of release years for Moviesvs TV Shows using seaborn's histplot()
    ### Use different color for Movies and TV shows


```python
df=df.dropna(subset=['type','release_year'])
movies=df[df['type']=='Movie']
shows=df[df['type']=='TV Show']
```

```python
sns.histplot(movies, shows,hue='type',palette='deep')
plt.show()
```
