# matplotlib.pyplot.subplots()
    ## subpllot(): provides away to plot multipple plot in single figure.
    ## Given number of rows and column. It returns a tuple (fig.ax)giving a single figure fig with an array of axes ax.

```python
import matplotlib.pyplot as plt
```

```python
fig,((ax1,ax2),(ax3,ax4))=plt.subplots(nrows=2,ncols=2,figsize=(10,5))
```

```python
import numpy as np
```

```python
sales_data={'Electronics':45000,'Clothing':30000,'Grocery':25000,'Mobile':20000}
category=list(sales_data.keys())
actual_sale=list(sales_data.values())
fig,((ax1,ax2),(ax3,ax4))=plt.subplots(nrows=2,ncols=2,figsize=(10,5))
x=np.random.random(10)
ax1.plot(x,x/2)
ax2.scatter(np.random.random(10),np.random.random(10))
ax3.bar(category,actual_sale)
ax4.hist(np.random.randn(1000))
```

## subplot using indexing

```python
our_lines=np.linspace(0,10,100)
```

```python
product_prices={'Hair oil':20,'Face Wash':30,'Lip Balm':10,'Face Cream':15}
```

```python
our_numbers=np.random.random(10)
```

```python
our_dots=np.random.randn(100)
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(10,5)) #setting subplots
#-------------------------------------------------------------------
ax[0,0].plot(our_lines,our_lines/2)
ax[0,1].bar(list(product_prices.keys()),list(product_prices.values()),color=['blue','green','red','orange'])
ax[1,0].scatter(np.random.randn(100),np.random.randn(100),color='red')
ax[1,1].hist(our_dots,color='yellow')
#----------------------------------------------------------
ax[0,0].set_title('Line Plot Example')
ax[0,0].set_xlabel('X-Axis')
ax[0,0].set_ylabel('Y-Axis')
ax[0,0].grid(color='grey',linestyle='--')
#-----------------------------------------------------------
ax[0,1].set_title('Product Sales Information')
ax[0,1].set_xlabel('Product Name')
ax[0,1].set_ylabel('Product Sale')
ax[0,1].grid(color='lightgreen',alpha=0.6,linestyle='--')
#------------------------------------------------------------
fig.suptitle('Matplotlib Subplots example using 2 rows and 2cols')
#------------------------------------------------------------
plt.tight_layout()
#plt.savefig('subplot.jpeg',dpi=300,bbox_inches='tight')
```
