```python
import numpy as np
import random
```

```python
#Define state and action
states = np.arange(0,5)
actions = ['GREEN','RED']
print('States: ', states)
print('Actions: ', actions)
```

```python
#Create Q table
#Q table :[s,a]
Q = np.zeros ((len(states),len(actions)))
alpha = 0.1    #learning rate
gamma = 0.9    #Discount factor
epsilon = 0.2  #Exploration rate
episodes = 300 #Training runs
print('Q-table shape :',Q.shape)
```

```python
#write rewards, penalty function
def get_reward(cond, action):
    if cond ==2:
        reward = +1 #ideal
    else:
        reward = 0 #too low/high
        
    if action == 'GREEN' and cond ==4 :
        reward += 10 #Clear Congestion
    if action == 'GREEN' and cond ==0 :
        reward -= 5 #waste power    
    if action == 'RED' and cond ==0 :
        reward += 5 #save energy
    if action == 'RED' and cond ==4 :
        reward -= 10 #Create Jam
    return reward
    
print('Reward Example (level=0, RED):',get_reward(4,'RED'))
```

```python
#Step 4 Enviroment Dynamic Change (e.g. Traffic change)
def next_cond(cond, action):
    if action == 'RED':
        cond += random.choice([1])
    else:
        cond -= random.choice([1])
    return int(np.clip(cond, 0, 4))

print('Next Condition Example: ', next_cond(0,'GREEN'))
```

```python
for ep in range(episodes):
    cond= random.choice(states)
    done= False
    for _ in range(20):
 
        if random.uniform(0, 1) < epsilon:
            action = random.choice(actions)
        else:
            action = actions[np.argmax(Q[cond//1])]
 
        next_state = next_cond(cond, action)
        reward = get_reward(next_state, action)
        a = actions.index(action)
        best_next = np.max(Q[next_state//1])
        Q[cond//1, a] += alpha * (reward + gamma * best_next - Q[cond//1, a])
 
       # cond = next_state

 
print('Training Done')
```
