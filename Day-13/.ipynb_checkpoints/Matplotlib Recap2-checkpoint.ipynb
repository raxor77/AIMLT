```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.random.rand(50)
y=np.random.rand(50)
temp=np.random.randint(20,100,size=50)
plt.scatter(x,y,c=temp,cmap='plasma',s=120,edgecolor='black') #cmap=plasma,inferno,cool,magma,inferno
plt.colorbar(label='Temperature')
plt.title('Scatter plot Color by Temperature')
plt.xlabel('X-Random')
plt.ylabel('Y-Random')
plt.show()
```

```python
height_cm=[150,155,160,165,170,175,180,185]
weight_kg=[50,53,56,59,62,66,68,72]
age_year=[20,22,25,28,30,35,38,40]
```

```python
plt.figure(figsize=(8,5))
```

```python
plt.scatter(height_cm,
            weight_kg,
            s=120,
            c=age_year,
            cmap='rainbow',
            marker='o',
            alpha=0.9)  #c=age_year refering to age data
plt.colorbar(label='Age(year)')
plt.title('Scatter plot Color by Temperature')
plt.xlabel('Height(cm)')
plt.ylabel('Weight(kg)')
plt.grid(color='blue',linestyle='--',linewidth=0.7)
plt.show()
```

## Pie Plot Example
    ### plt.pie(x,labels='',colors='',autopct='',startangle='',shadow=,explode=,radius=)

```python
depts=['HR','Web-Development','App-Development','Agile-Scrum','QA']
expenses=[12000,25000,18000,15000,20000]
```

```python
plt.figure(figsize=(7,7))
```

```python
plt.pie(expenses,
        labels=depts,
        autopct='%1.2f%%',
        startangle=140,
        explode=(0,0.05,0,0.05,0),
        shadow=True)   
plt.title('Departments Expenses')
plt.show()
```

```python
plt.pie(expenses,
        labels=depts,
        colors=plt.cm.plasma([0.0,0.2,0.5,0.7,0.9]),
        autopct='%1.0f%%',
        startangle=140,
        explode=(0,0.05,0,0.05,0),
        radius=1.1,
        shadow=True)   #autopct for point #cool can be rainbow,hot,cool,plasma,inferno,magma
plt.title('Departments Expenses')
plt.show()
```

## Create pie chart using following Data
    ### sales_data={'Mobile':45000,'Clothing':30000,'Groceries':35000,'Home Decor':2000}

```python
sales_data={'Mobile':45000,'Clothing':30000,'Groceries':35000,'Home Decor':20000}
department=list(sales_data.keys())
ttl_sales=list(sales_data.values())
plt.figure(figsize=(10,10))
plt.pie(ttl_sales,
        labels=department,
        colors=plt.cm.cool([0.2,0.5,0.7,0.9]),
        autopct='%1.2f%%',
        startangle=140,
        explode=(0,0,0,0.05),
        textprops={'color':'indigo','fontsize':10,'fontweight':'bold'},
        shadow=True)
plt.title('Product Sales by Dept.')
# plt.savefig('categorysale.png',dpi=300,bbox_inches='tight')
plt.show()
```

## Histogram Example

```python
marks=[10,20,30,35,45,60,55,62,0,90,89,50,56,72,80,91,86,55,99,60,83,100]
plt.style.use('ggplot')
plt.figure(figsize=(8,5))
plt.hist(marks,bins=5,color='green',edgecolor='darkblue',alpha=0.8) #bin change number of bars
plt.title('Histogram for Student Marks')
plt.xlabel('Marks Range')
plt.ylabel('Number of Student')
plt.grid(color='blue',linestyle='--',linewidth=0.7,alpha=0.3)
plt.show()
```
