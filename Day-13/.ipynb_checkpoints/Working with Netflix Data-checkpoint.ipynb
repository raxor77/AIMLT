```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('netflix_data.csv')
display(df.head())
```

```python
df=pd.read_csv('netflix_data.csv')
df=df.dropna(subset=['release_year'])
df['release_year']=df['release_year'].astype(int)
plt.figure(figsize=(8,5))
plt.hist(df['release_year'],bins=100,color='green',edgecolor='yellow',alpha=0.8) #bin change number of bars
plt.title('Netflix Release Title by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(color='blue',linestyle='--',linewidth=0.7,alpha=0.3)
plt.tight_layout()
# plt.savefig('Netflix_bar.pdf',dpi=300,bbox_inches='tight')
plt.show()
```

```python
df=df.dropna(subset=['release_year','type'])
```

```python
df['release_year']=df['release_year'].astype(int)
```

```python
movies=df[df['type']=='Movie']
shows=df[df['type']=='TV Show']
```

```python
plt.figure(figsize=(10,5))
plt.hist([movies['release_year'],
         shows['release_year']],
         bins=30,
         color=['green','blue'],
         edgecolor='yellow',
         alpha=0.8,
        label=['Movies','TV-Shows'])
plt.title('Netflix Release Title by Year and Type')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles Released')
plt.grid(color='blue',linestyle='--',linewidth=0.7,alpha=0.3)
plt.tight_layout()
plt.legend(title='Movies/Shows')
# plt.savefig('Netflix_Movies_Shows.pdf',dpi=300,bbox_inches='tight')
plt.show()
```

### print how many Movies and TV Shows are available on Netflix
### Display Movies and TV Shows propotion in a pie plot
### Save as pdf

```python
import numpy as np
```

```python
df=pd.read_csv('netflix_data.csv')
df=df.dropna(subset=['type'])  #to clean data from null
type_counts=df['type'].value_counts()
print(type_counts)
plt.figure(figsize=(8,8))     
plt.pie(type_counts,
        labels=type_counts.index,
        colors=plt.cm.cool([0.2,0.9]),
        autopct='%1.2f%%',
        startangle=140,
        explode=(0.05,0.05),
        shadow=True)
plt.title('Netflix Movies and TV Shows Distribution')
plt.legend(title='Movies and TV Show')
#plt.savefig('Netflix plot Type.pdf',dpi=300,bbox_inches='tight')
plt.show()
```
