# Matplotlib recap

```python
import matplotlib.pyplot as plt
```

```python
months=['jan','feb','mar','april']
sales=[2000,3000,2500,3600]
```

```python
plt.plot(months,sales,color='hotpink',linestyle='--',linewidth=2,marker='p',label='Monthly Sales')
#plt.figure(figsize=(7,7))  #cannot use in plot
plt.title('Monthly Sales Report')
plt.xlabel('Month')
plt.ylabel('Sales Total')
plt.xlim(0,4)  #x axis section
plt.ylim(2000,4000) #y axis section
plt.grid(True,color='gray',linestyle='--') #grid color also style
plt.legend(loc='lower right') #need to add label='***' in plt.plot
plt.show()
```

```python
import numpy as np
```

```python
x=np.linspace(0,2,100)
plt.plot(x,x,label='Linear')
plt.plot(x,x**2,label='Quadratic')
plt.plot(x,x**3,label='Cubic')
plt.xlabel('x Lines')
plt.ylabel('Y lines')
plt.title('Simple Lines Plot')
plt.legend()
plt.show()
```

## Bar Plot example
    ### plt.bar(x,y,colors="",edgecolor="")

```python
categories=['Electronics','MObiles','Books','Groceries','Home Decoration']
sales=[50000,60000,30000,90000,40000]
bar_col=['red','blue','green','cyan','orange']
plt.figure(figsize=(8,5))
plt.bar(categories,sales,color=bar_col,edgecolor='black')
plt.xlabel('Categories')
plt.ylabel('Sales Amount (MYR)')
plt.title("Category Sales Report")
plt.tight_layout()
plt.grid(axis='y',color='gray',alpha=0.3,linestyle='--') #alpha is transparentsy #True will show both x,y axis grid
plt.show()
```

```python
categories=['Electronics','MObiles','Books','Groceries','Home Decoration']
sales=[50000,60000,30000,90000,40000]
plt.figure(figsize=(8,5))
plt.bar(categories,sales,color='blue',edgecolor='black',label='Sales Info')
plt.xlabel('Categories')
plt.ylabel('Sales Amount (MYR)')
plt.title("Category Sales Report")
plt.grid(axis='y',color='gray',alpha=0.3,linestyle='--') #alpha is transparentsy #True will show both x,y axis grid
plt.xticks(rotation=30,fontsize=12,color='magenta') #change xtick display
plt.yticks([20000,40000,60000,80000,100000],('20k','40k','60k','80k','1M'),fontsize=12,color='green') #change ytick display
plt.legend(title='Legend',fontsize=8)
plt.tight_layout()
plt.show()
```

## Scatter Chart Example
    ### plt.scatter(x,y,s='',c='colorName',marker='makername',alpha=tranp., edgecolor='colorname',label='label of legend')

```python
height_cm=[150,155,160,165,170,175,180,185]
weight_kg=[50,53,56,59,62,66,68,72]
plt.figure(figsize=(8,5))
plt.scatter(height_cm,
            weight_kg,
            s=100,
            c='green',
            marker='x',
            alpha=0.8,
            label='Weight Height comparison')
plt.title('Height vs Weight Scatter Chart')
plt.legend()
plt.xlabel('Height(cm)')
plt.ylabel('Weight (kg)')
plt.xticks(rotation=45,fontsize=10,color='brown') #change xtick display
plt.yticks([50,55,60,65,70,75,80,85],('50kg','55kg','60k','65kg','70kg','75kg','80kg','85kg'),fontsize=10,color='green')
plt.grid(color='lightblue',alpha=0.8,linestyle='--',linewidth=0.5)
plt.tight_layout()
plt.show()
```
