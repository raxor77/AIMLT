```python
import pandas as pd
```

```python
#Sample dataset
data = {
    'Name' : ['ali','sarah','johan',None,'zaid'],
    'Age' : [25,None,30,22,None],
    'City' :['KL','Penang',None,'Johor','KL']
}
```

```python
df=pd.DataFrame(data)
print('Original Data Frame :')
print(df)
```

```python
#check how many missing
df.isnull().sum()
```

```python
#Check missing in row
for index, row in df.iterrows():
    print(f'Row {index} : Missing ={row.isnull().any()}, Details={row.isnull().to_dict()}')
```

```python
#Show only rows with missing data
missing_rows = df[df.isnull().any(axis=1)]
print('Rows with missing data')
print(missing_rows)
print('Above rows will be drop using dropna()')
```

```python
#cleaned data with dropna()
df_cleaned = df.dropna()
print('nAfter dropna() data')
print(df_cleaned)
```

```python
print('Before fillna()')
print(df)
```

```python
df_filled = df.fillna({'Name':'Unknown',
                       'Age' : df['Age'].mean(),
                       'City': 'Not Available'})
#Data display after fillna()
print('\nAfter fillna()') 
print(df_filled)
```

```python
#round at 1 decimal
df_fills = df.fillna({'Name':'Unknown',
                       'Age' : round(df['Age'].mean(),1),
                       'City': 'Not Available'})
print('\nAfter fillna() : 2 decimal')
print(df_fills)
```

```python
import math
```

```python
df_floor = df.fillna({
    'Name': 'Unknown',
    'Age': math.floor(df['Age'].mean().astype(int)),
    'City': 'Not Available'
})
print('\nAfter fillna() : a floor')
print(df_floor)
```

```python
df_new = df.fillna({
    'Name': 'Unknown',
    'Age': math.ceil(df['Age'].mean().astype(int)),
    'City': 'Not Available'
})
print('\nAfter fillna() : a ceilling')
print(df_new)
```
