### First sample to create a line chart form or Data
    ### students.xlsx for fetchhing the data
    ### generate Line chart base on students name and Marks

```python
import pandas as pd
```

```python
df=pd.read_excel('C:/Users/User/AIML/Day-11/students.xlsx')
```

```python
display(df)
```

```python
names=df['Name']
marks=df['Marks']
```

```python
plt.plot(names,marks,marker='s',color='green')
plt.title('Student Marks Line Chart')
plt.xlabel('Student Names')
plt.ylabel('Student Marks')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(5,3))
plt.plot(names,marks,marker='>',color='orange')
plt.title('Student Marks Progress Line Chart')
plt.xlabel('Names')
plt.ylabel('Marks')
plt.grid(True)
plt.show()
```

```python
print('Dotted line')
plt.figure(figsize=(5,3))
plt.plot(names,marks,marker='>',color='orange',linestyle='-')
plt.title('Student Marks Progress Line Chart')
plt.grid(True)
plt.show()
```
