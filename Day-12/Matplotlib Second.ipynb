```python
import pandas as pd
df=pd.read_csv('patients.csv')
print(df.head())
```

```python
age_counts=df['age'].value_counts().sort_index()
plt.figure(figsize=(15,10))
plt.bar(age_counts.index,age_counts.values,color='lavender',width=0.5)
plt.title('Number of Patients by Age')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.show()
```

# Create Bar plot-Average Age of Patients by Arrival date

```python
print(df.head(0))
```

```python
df['arrival_date']=pd.to_datetime(df['arrival_date'])                         
avg_age_by_date=df.groupby('arrival_date')['age'].mean()
plt.figure(figsize=(15,10))
plt.bar(avg_age_by_date.index.astype(str),avg_age_by_date.values,color='blue',width=0.5)
plt.title('Average Age of Patients by Arrival date')
plt.xlabel('Arrival Date')
plt.ylabel('Average Age')
plt.show()
```

```python
patients_by_service=df.groupby('service')['age'].mean()
print(patients_by_service)
plt.figure(figsize=(6,5))
plt.bar(patients_by_service.index,patients_by_service.values,color='blue',width=0.5)
plt.title('Average Age of Patients by service')
plt.xlabel('Service')
plt.ylabel('Total of Patients')
plt.show()
```

## Scatter Plot
### Scatter Plot represent data in points, where each point represent a Data observation, we can check how to values correlate

```python
products_list={'Laptop':750,'smartphone':650,'smartwatch':400,'tablets':500,'AC':400}
all_products=list(products_list.keys())
ttl_sales=list(products_list.values())
plt.scatter(all_products,ttl_sales)
plt.title('Basic product and Sales Representation')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
players=['anuar','sabri','muhiddin','mahathir','najib']
match1_score=[45,67,56,78,89]
match2_score=[55,78,40,80,97]
match3_score=[50,75,40,82,89]
plt.figure(figsize=(8,5))
plt.scatter(players,match1_score,color='red',label='Match One',s=100)
plt.scatter(players,
             match2_score,
             color='yellow',
             label='Match Two',
             s=100,
             marker='s')
plt.scatter(players,
             match3_score,
             color='green',
             label='Match Two',
             s=100,
             marker='d')
plt.title('Comparison of Players in 2 Matches')
plt.xlabel('Players')
plt.ylabel('Scores')
plt.legend(loc='lower right')
plt.grid(True,
        linestyle='--',
        alpha=0.5)
plt.show()
```

# Write a comparisonthat shows average calories burned for each workout type from final data data set

```python
df=pd.read_csv('final_data.csv')
avg_cal=df.groupby('Workout_Type')['Calories_Burned'].mean()
plt.figure(figsize=(6,5))
fg,ax=plt.subplots()  #backgroud colour change
ax.set_facecolor('lightyellow') #backgroud colour change
plt.scatter(avg_cal.index,
            avg_cal.values,
            s=100,marker='p',
            color='blue',
            facecolor='cyan')
plt.title('Average Calories Burned for each WorkOut')
plt.xlabel('Workout Type')
plt.ylabel('Average Calories Burned')
plt.grid(True,
         linestyle="--",
         color='grey')
plt.show()
```
